```python
import torch
import os
from PIL import Image

# Load the model
model = torch.hub.load('.', 'custom', path='ArchTrain.pt', source='local') 

# Directory containing images
img_dir = 'images/test/'
image_extensions = ['.jpg', '.jpeg', '.png']  # Add other extensions if needed

# Iterate through each file in the directory
for img_file in os.listdir(img_dir):
    if any(img_file.lower().endswith(ext) for ext in image_extensions):
        img_path = os.path.join(img_dir, img_file)
        # Run inference
        results = model(img_path)
        # Show the results
        results.show()

print("Inference completed on all images.")
```

```python
import torch
import cv2
import os
import pandas as pd

# Load the model
model = torch.hub.load('.', 'custom', path='ArchTrain.pt', source='local')

# Load image
img_path = 'images/test/Cat24_7.jpg'
img = cv2.imread(img_path)

# Inference
results = model(img)

# Create a directory to save cropped images
output_dir = 'cropped_images'
os.makedirs(output_dir, exist_ok=True)

# List to hold data for CSV
data = []

# Iterate over the detections
for i, (xmin, ymin, xmax, ymax, conf, cls) in enumerate(results.xyxy[0]):
    # Convert bounding box coordinates to integers
    x1, y1, x2, y2 = map(int, (xmin, ymin, xmax, ymax))

    # Crop the image
    cropped_img = img[y1:y2, x1:x2]

    # Create a filename with coordinates
    cropped_img_filename = f'cropped_x{x1}_y{y1}_width{x2-x1}_height{y2-y1}.jpg'
    cropped_img_path = os.path.join(output_dir, cropped_img_filename)
    
    # Save the cropped image
    cv2.imwrite(cropped_img_path, cropped_img)

    # Append data to the list for CSV
    data.append({'filename': cropped_img_filename, 'x': x1, 'y': y1})

    print(f'Cropped image saved at: {cropped_img_path}')

# Create a DataFrame from the data and save to CSV
df = pd.DataFrame(data)
csv_path = os.path.join(output_dir, 'coordinates.csv')
df.to_csv(csv_path, index=False)

print(f'Coordinates saved to: {csv_path}')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# If using Jupyter Notebook, uncomment the following line
%matplotlib inline 

# Load the CSV file containing the coordinates and filenames
csv_path = 'cropped_images/coordinates.csv'
data = pd.read_csv(csv_path)

# Create a blank figure with the specified size (same as original image size)
fig, ax = plt.subplots(figsize=(12.81, 12.81))  # 1281 x 1197 pixels

# Iterate over each row in the DataFrame
for index, row in data.iterrows():
    # Load the cropped image
    img_path = f'cropped_images/{row["filename"]}'
    img = mpimg.imread(img_path)

    # Get the x and y coordinates
    x, y = row['x'], row['y']
    
    # Debugging output
    print(f"Processing image: {row['filename']} at coordinates (x: {x}, y: {y}) with shape {img.shape}")

    # Display the image at the specified coordinates
    ax.imshow(img, aspect='auto', extent=(x, x + img.shape[1], y, y + img.shape[0]))

# Set limits for x and y axes (optional)
ax.set_xlim(0, 1281)  # Set to original image width
ax.set_ylim(0, 1197)  # Set to original image height
ax.set_axis_off()  # Hide the axes

# Show the plot
plt.show()
```
